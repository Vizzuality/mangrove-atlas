```python
import pandas as pd
import os
import geopandas as gpd
```

```python
DATA_DIR = ('../../../../data/GMW_extent_V4')
```

## Load locations

```python
locations_combined = pd.read_csv('https://storage.googleapis.com/mangrove_atlas/widget_data/locations_merged.csv')
locations_combined.head()
```

```python
locations_combined['type'].value_counts()
```

```python
locations_country = locations_combined[locations_combined['type'] == 'country']
locations_wdpa = locations_combined[locations_combined['type'] == 'wdpa']
locations_wdpa['wdpaid'] = locations_wdpa['wdpaid'].astype(int)
```

## Extent data

### Countries

```python
country_stats = pd.read_parquet(os.path.join(DATA_DIR,
                                             'gmw_mng_extent_stats_v4112_corrected',
                                             'gmw_v4_timeseries_4112_gmw_country_stats_corr_area.parquet'))
country_stats.reset_index(drop=False, inplace=True)
#Correct ISO codes
country_stats.loc[country_stats['cnty_name'] == 'Bonaire', 'iso'] = 'BES'
country_stats.loc[country_stats['cnty_name'] == 'Europa Island', 'iso'] = 'ATF'
print(len(country_stats))
country_stats.head(10)
```

```python
#Extent locations without match on locations
country_stats[~country_stats['iso'].isin(locations_country['iso'])]
```

```python
locations_country[~locations_country['iso'].isin(country_stats['iso'])]
```

```python
country_stats_id = locations_country[['iso', 'id']].merge(country_stats, how='left', left_on='iso', right_on='iso')
country_stats_id.drop(columns=['iso', 'cnty_name'], inplace=True)
country_stats_long = country_stats_id.melt(id_vars=['id'], var_name='year', value_name='value')
country_stats_long['indicator'] = 'habitat_extent_area'
country_stats_long = country_stats_long[['id','indicator', 'year',  'value']]
country_stats_long.sort_values(by=['id', 'year'], inplace=True)
country_stats_long.dropna(inplace=True)

country_stats_long.head(10)
```

### WDPAs

```python
wdpa_stats = pd.read_parquet(os.path.join(DATA_DIR,
                                         'gmw_v4112_stats_wdpa_wdoecms',
                                         'mng_ext_stats_v4112',
                                         'gmw_v4_timeseries_v4112_wdpa_jun2026_stats_corr_area.parquet'))
wdpa_layer = gpd.read_file(os.path.join(DATA_DIR,
                                        'gmw_v4112_stats_wdpa_wdoecms',
                                        'WDPA_poly_Jun2026_intersect_hab_mask_v29_uid_n_pxls_vld.gpkg'))
wdpa_layer = wdpa_layer[['SITE_ID']]

wdpa_stats['wdpaid'] = wdpa_layer['SITE_ID'].values # !! direct merge by order of rows in the geopackage and parquet file
wdpa_stats.head(3)
```

```python
print("WDPA locations in database", len(locations_wdpa))
print("WDPA locations in stats", len(wdpa_stats))
print("Database locations without match on stats", len(locations_wdpa[~locations_wdpa['wdpaid'].isin(wdpa_stats['wdpaid'])]))
```

```python
wdpa_stats_id = locations_wdpa[['wdpaid', 'id']].merge(wdpa_stats, how='left', left_on='wdpaid', right_on='wdpaid')
wdpa_stats_id.drop(columns=['wdpaid', 'name'], inplace=True)

wdpa_stats_long = wdpa_stats_id.melt(id_vars=['id'], var_name='year', value_name='value')
wdpa_stats_long['indicator'] = 'habitat_extent_area'
wdpa_stats_long = wdpa_stats_long[['id','indicator', 'year',  'value']]
wdpa_stats_long.sort_values(by=['id', 'year'], inplace=True)
wdpa_stats_long.dropna(inplace=True)
wdpa_stats_long.head(10)
```

```python
len(wdpa_stats_long), len(country_stats_long)
```

```python
extent_stats_long = pd.concat([country_stats_long, wdpa_stats_long], axis=0)
extent_stats_long.rename(columns={'id': 'location_id'}, inplace=True)
#convert values from ha to km2!
extent_stats_long['value'] = extent_stats_long['value'] * 0.01

extent_stats_long.to_csv(os.path.join('../../../../data', 'extent_stats_v4.csv'), index=False)
#extent_stats_long[0:3].to_csv(os.path.join('../../../../data', 'extent_stats_v4_TEST.csv'), index=False)
```
